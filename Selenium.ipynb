```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.chrome.options import Options
import time
import pandas as pd


options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")
options.add_experimental_option("excludeSwitches", ["enable-automation"])
options.add_experimental_option('useAutomationExtension', False)
#options.add_argument("--headless=new")
options.add_argument("--window-size=1920,1080")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")


driver = webdriver.Chrome(options=options)


list_of_jobs = []

driver.get("https://www.naukrigulf.com/data-engineer-jobs")
time.sleep(5)


for i in range (3):
    jobs = driver.find_elements(By.CLASS_NAME, 'ng-box')
    for job in jobs:
        try:
            job_title = job.find_element(By.CLASS_NAME, 'designation-title').text
        except:
            job_title = "None"

        try:
            job_company = job.find_element(By.CLASS_NAME, 'info-org').text
        except:
            job_company = "None"

        try:
            job_location = job.find_element(By.XPATH, '//*[@id="searchResult"]/div/section[2]/div[2]/div[2]/div[1]/ul/li[2]/span[2]').text
        except:
            job_location = "None"

        try:
            job_url = job.find_element(By.XPATH, 'a').get_attribute('href')
        except:
            job_url = "None"

        try:
            job_experience = job.find_element(By.XPATH, '//*[@id="searchResult"]/div/section[2]/div[2]/div[2]/div[1]/ul/li[1]/span[2]').text
        except:
            job_experience = "None"

        try:
            job_description = job.find_element(By.CLASS_NAME, 'description').text
        except:
            job_description = "None"

        list_of_jobs.append({
            'job_title': job_title, 
            'job_company': job_company,
            'job_location': job_location,
            'job_description': job_description,
            'job_experience': job_experience,
            'job_url': job_url
       })
    time.sleep(8)
    if i < 2:
        time.sleep(5)
        if i == 0:
            driver.find_element(By.CSS_SELECTOR, ".arr-wrapper").click()
        else:
            driver.find_element(By.CSS_SELECTOR, ".ico.bwd").click()

    time.sleep(5) 

df = pd.DataFrame(list_of_jobs)
df.drop_duplicates(subset=['job_url'], inplace=True)
df.to_csv('Jobs.csv', index=False)
```
